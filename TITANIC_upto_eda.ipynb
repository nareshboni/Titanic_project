```python
from scipy import stats

# Calculate z-scores
z_scores = stats.zscore(df[['Age', 'Fare']])

# Boolean mask for outliers
outliers = (abs(z_scores) > 3).any(axis=1)

# View outlier rows
df[outliers]  # ✅ This works

# OR, add outliers as a new column
df['outliers'] = outliers  # ✅ This now adds a column to the DataFrame
```

```python
df.head()
```

```python
## function to rempove outlier using IQR
def remove_outlier_iqr(df,column):
    Q1=df[column].quantile(0.25)
    Q3=df[column].quantile(0.75)
    IQR=Q3-Q1
    lower=Q1-1.5*IQR
    upper=Q1+1.5*IQR
    return df[(df[column] >= lower) & (df[column] <= upper)]


# Remove outliers from each numerical column
for col in numerical_features:
    df = remove_outlier_iqr(df, col)
```
